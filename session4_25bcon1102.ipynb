```python
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks


s1 = Student("ADESH", 101, "60")

print("Name:", s1.name)
print("Roll:", s1.roll)
print("Marks:", s1.marks)
```

```python
ef find_average(numbers):
    total = 0
    for i in range(len(numbers) - 1):
        total = total + numbers[i]
    return total / len(numbers)

result = find_average([10, 20, 30, 40])
print(result)
```
